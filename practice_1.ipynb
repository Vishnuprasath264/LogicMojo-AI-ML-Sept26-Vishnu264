```python
# Q1. Age Validation
# Take age as input from the user. If the input is empty, raise a ValueError. If the age is negative, raise another ValueError. Use try-except to handle the errors. If the age is 18 or above, print "Eligible to vote"; otherwise, print "Not eligible to vote".
try:
    age = int(input("Enter your age:"))
    if age < 0:
        raise(ValueError)
    if age is None:
        raise(ValueError)
    if age >= 18:
        print("Eligible to Vote")
    else:
        print("Not Eligible to Vote")
except ValueError:
    print("Age is invalid")
```

```python
# Q2. Positive Number Checker
# Ask the user to enter a number. Handle invalid numeric input using try-except. If the number is zero or negative, raise a ValueError. If it is positive, check whether it is even or odd and display the result.
try:
    age = int(input("Enter a number:"))
    if age <= 0:
        raise(ValueError)
    if age % 2 == 1:
        print("The Number is odd")
    else:
        print("The Numer is even")
except ValueError:
    print("The Number is invalid")
```

```python
# Q3. Username Validation
# Take a username as input. If the username is empty or contains fewer than 5 characters, raise a ValueError. Use try-except to handle the exception. If the username is valid, print "Username accepted".
try:
    username = input("Enter a username:")
    if username is None or len(username) < 5:
        raise(ValueError)
    print("Username Accepted")
except ValueError:
    print("Username is invalid")
```

```python
# Q4. Division Calculator
# Take two numbers as input. Handle invalid input using try-except. If the second number is zero, raise a ZeroDivisionError. Otherwise, divide the first number by the second and print the result. Use else to display the result only when no exception occurs.
try:
    a = int(input("Enter a number:"))
    b = int(input("Enter a number:"))
    if b == 0:
        raise(ZeroDivisionError)
    else:
        print(a/b)
except ValueError:
    print("Number is Invalid")
except ZeroDivisionError:
    print("Second number cannot be zero")
```

```python
# Q5. Student Marks Validation
# Take marks as input. If the input is not a valid integer, handle the exception. If marks are less than 0 or greater than 100, raise a ValueError. Otherwise, use if-else to print "Pass" if marks are at least 40 and "Fail" otherwise.
try:
    marks = int(input("Enter the mark:"))
    if marks < 0 or marks > 100:
        raise(ValueError)
    if marks > 40:
        print("Pass")
    else:
        print("Fail")
except ValueError:
    print("Invalid Marks")
```

```python
# Q6. Create a Student File
# Use "x" mode to create a file named student.txt. Take the student's name, age, and course as input and write these details into the file. Handle FileExistsError if the file already exists.
try:
    file = open("student.txt","x")
    s_name = input("Enter student's name:")
    s_age = int(input("Enter student's age:"))
    if s_age <= 0:
        raise(ValueError)
    s_course = input("Enter student's course:")
    file.write(f"Name: {s_name}")
    file.write(f"\nAge: {s_age}")
    file.write(f"\nCourse: {s_course}")
    file.close()
except FileExistsError:
    print("File already exists")
    file.close()
except ValueError:
    print("Invalid Value")
    file.close()
```

```python
# Q7. Write Student Records
# Use "w" mode to create or overwrite students.txt. Take the details of three students and write each student's name and marks on a separate line. Handle any appropriate input errors using try-except.
try:
    file = open("students.txt","w")
    for i in range(2):
        name = input("Enter student name:")
        marks = int(input("Enter student marks:"))
        if marks < 0 or marks >100:
            raise(ValueError)
        file.write(f"{name}: {marks}\n")
    file.close()
except ValueError:
    print("Invalid Marks")
    file.close()
```

```python
# Q8. Append Student Records
# Use "a" mode to add a new student's name and marks to the existing students.txt file. If the marks are below 0 or above 100, raise a ValueError. Handle the exception without adding invalid records to the file.
try:
    file = open("students.txt","a")
    name = input("Enter student name:")
    marks = int(input("Enter student marks:"))
    if marks < 0 or marks >100:
        raise(ValueError)
    file.write(f"{name}: {marks}\n")
    file.close()
except ValueError:
    print("Invalid Marks")
    file.close()
```

```python
# Q9. Read a File Safely
# Ask the user to enter a filename. Open the file using "r" mode and display its contents. Handle FileNotFoundError if the file does not exist. If the file is empty, print "File is empty".
try:
    filename = input("Enter the file name:")
    if filename[-4::] != ".txt":
        filename += ".txt"
    file = open(filename,"r")
    info = file.read()
    if info is None:
        print("File is empty")
    else:
        print(info)
    file.close()
except FileNotFoundError:
    print("File does not exist")
    file.close()
```

```python
# Q10. Number Storage in a File
# Take five numbers from the user and write them to numbers.txt using "w" mode. If a user enters an invalid number, handle the exception. If a number is negative, raise a ValueError. Store only valid, non-negative numbers in the file.
try:
    file = open("numbers.txt","w")
    for i in range(5):
        num = int(input("Enter a number:"))
        if num < 0:
            raise(ValueError)
        file.write(f"{num},")
    file.close()
except ValueError:
    print("The number is invalid(enter non-negative numbers only)")
    file.close()
```

```python
# Q11. Search for a Student
# Read students.txt using "r" mode. Ask the user to enter a student name. If the name exists in the file, print "Student found"; otherwise, print "Student not found". Handle FileNotFoundError if the file does not exist.
try:
    file = open("students.txt","r")
    student = file.read()
    name = input("Enter the student's name:")
    if name in student:
        print("Student found")
    else:
        print("Student not found")
except FileNotFoundError:
    print("File does not exist")
```

```python
# Q12. File Creation with Validation
# Ask the user to enter a filename. If the filename is empty, raise a ValueError. Otherwise, use "x" mode to create the file. Handle both ValueError and FileExistsError appropriately.
try:
    filename = input("Enter a valid filename:")
    if filename[-4::] != ".txt":
        filename += ".txt"
    if filename == "":
        raise(ValueError)
    file = open(filename,"x")
    file.close()
except ValueError:
    print("Filename should not be empty")
except FileExistsError:
    print("Filename is already used")
```

```python
# Q13. Append Only Passing Students
# Take a student's name and marks as input. If marks are outside the range 0–100, raise a ValueError. If marks are 40 or above, append the student's details to passed_students.txt using "a" mode. Otherwise, print "Student failed; record not added".
try:
    file = open("passed_students.txt","a")
    name = input("Enter student name:")
    marks = int(input("Enter student marks:"))
    if marks < 0 or marks > 100:
        raise(ValueError)
    if marks >= 40:
        file.write(f"{name}: {marks}\n")
        print("Record added")
    else:
        print("Record not added")
    file.close()
except ValueError:
    print("Marks should be between 0 and 100")
    file.close()
```

```python
# Q14. Count Lines in a File
# Ask the user for a filename and read it using "r" mode. Count the number of non-empty lines in the file. Handle FileNotFoundError. If the file contains no non-empty lines, print "No records found".
try:
    filename = input("Enter a file name:")
    if filename[-4::] != ".txt":
        filename += ".txt"
    file = open(filename, "r")
    count = 0
    for line in file:
        print(line)
        count += 1
    if count == 0:
        print("No records found")
except FileNotFoundError:
    print("File does not exist")
```
