```python
#55. Write a decorator that prints "Training started" before a function executes and "Training finished" after it executes. Apply it to a function that prints the current epoch.
def decorator(func):
    def wrapper(*args):
        print("Training started")
        func(*args)
        print("Training finished")
    return wrapper

@decorator
def curr_epoch(total_epochs):
    for i in range(total_epochs):
        print("Current epoch:",i+1)

curr_epoch(5)
```

```python
#56. Write a decorator that accepts a function, executes it, and returns its result. Demonstrate it using a function that calculates model accuracy.
def decorator(func):
    def wrapper(*args):
        print("Calculating")
        func(*args)
        print("Done!")
    return wrapper

@decorator
def model_acc(correct, total):
    result = correct / total * 100
    if correct > total or result < 0:
        print("Invalid Result")
        return
    print("Model Accuracy:",result)

model_acc(25,100)
model_acc(-20,100)
```

```python
#60. Build a mini training simulation using functions, a decorator, a loop, conditions, break, and a lambda function. Print the loss for each epoch, stop when loss reaches 0.10 or lower, and return the final loss. Use only the Python concepts covered in class.
def decorator(func):
    def wrapper(*args):
        print("Training Started!")
        func(*args)
        print("Training Finished!")
    return wrapper

epoch = lambda value, loss: value - loss

@decorator
def model_sim(lv, epochs, loss):
    for i in range(epochs):
        lv = epoch(lv, loss)
        if lv < 0.1:
            break
        print(f"Loss value after Epoch {i+1} is {lv}.")
        
while True:
    loss_value = float(input("Enter the Loss value(-1 for exit):"))
    if loss_value == -1:
        break
    total_epochs = int(input("Enter the Total No. of Epochs(-1 for exit):"))
    if total_epochs == -1:
        break
    reduction = float(input("Enter the loss during each epoch(-1 for exit):"))
    if reduction == -1:
        break
    model_sim(loss_value, total_epochs, reduction)
```
